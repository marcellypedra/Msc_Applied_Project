```python
#Selecting chosen features for analysis

df = pd.read_csv("adult23.csv", usecols=[
    'ANXEV_A', 'AGEP_A', 'SEX_A', 'HISPALLP_A',
    'EDUCP_A', 'MLTFAMFLG_A', 
    'PCNTADLT_A', 'PCNTKIDS_A', 'OVER65FLG_A', 
    'MARSTAT_A', 'CITZNSTP_A'
])

print(df.shape)

df.head()
```

```python
#Managing null-vales for the analysis

import numpy as np

invalid_map = {
    'ANXEV_A': [7, 8, 9],
    'AGEP_A': [7, 8, 9],
    'SEX_A': [7, 8, 9],
    'HISPALLP_A': [97, 98, 99],
    'EDUCP_A': [97, 98, 99],  
    'MLTFAMFLG_A': [7, 8, 9],
    'PCNTADLT_A': [7, 8, 9],
    'PCNTKIDS_A': [7, 8, 9],
    'OVER65FLG_A': [7, 8, 9],
    'MARSTAT_A': [9, 97, 98, 99],  
    'CITZNSTP_A': [7, 8, 9]
}

for col, codes in invalid_map.items():
    df[col] = df[col].replace(codes, np.nan)
```

```python
#Checking for Unique values in columns

for col in df.columns:
    print(f"\nUnique values in '{col}':")
    print(df[col].sort_values().unique())
```

```python
#Turning all numbers in integer

exclude_cols = ['MLTFAMFLG_A', 'OVER65FLG_A', 'PCNTADLT_A', 'PCNTKIDS_A', 'HISPALLP_A', 'AGEP_A']

for col in df.columns:
    if col not in exclude_cols:  
        df[col] = df[col].astype('Int64')

for col in df.columns:
    print(f"\nUnique values in '{col}':")
    print(df[col].sort_values().unique())
```

```python
# Checking for missing values

missing_report = df.isnull().sum()

missing_report = missing_report[missing_report > 0]

print("Columns with missing values:")
print(missing_report)
```

```python
# Drop rows with missing data

df.dropna(inplace=True)

# Checking for missing values

missing_report = df.isnull().sum()

missing_report = missing_report[missing_report > 0]

print("Columns with missing values:")
print(missing_report)
```

```python
#Chechking for Selected Dataframe Info

print("Selected DataFrame Information:")
df.info()
```

```python
#Saving cleaned csv file
df.to_csv("nhis2023_cleaned.csv", index=False)
```

```python
# 

import matplotlib.pyplot as plt
import pandas as pd

# Step 1: Mapping Dictionaries from NHIS-2023 Codebook and creating readable labels
map_anxev = {1: "Yes", 2: "No"}
map_sex = {1: "Male", 2: "Female"}
map_race = {
    1: "Hispanic", 
    2: "Non-Hispanic White only",
    3: "Non-Hispanic Black/African American only", 
    4: "Non-Hispanic Asian only",
    5: "Non-Hispanic AIAN only",
    6: "Non-Hispanic AIAN and any other group",
    7: "Multiple races"
}

df['ANXEV_A'] = df['ANXEV_A'].replace(map_anxev)
df['SEX_A'] = df['SEX_A'].replace(map_sex)
df['HISPALLP_A'] = df['HISPALLP_A'].replace(map_race)

# Step 2: Function to Plot Counts with Percentages
def plot_with_percentages(counts, percentages, title, xlabel):
    plt.figure(figsize=(6, 4))
    bars = counts.plot(kind='bar', color='skyblue', edgecolor='black')
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--', alpha=0.7)

    # Add percentage labels
    for i, count in enumerate(counts):
        pct = percentages.iloc[i]
        plt.text(i, count + (count * 0.01), f"{pct:.1f}%", ha='center', va='bottom', fontsize=10)

    plt.show()

# ===== Step 3: Individual Distributions =====
cols_to_check = ['ANXEV_A', 'SEX_A', 'RACEALLP_A', 'HISP_A']

for col in cols_to_check:
    counts = df[col].value_counts()
    percentages = df[col].value_counts(normalize=True) * 100
    print(f"\n--- {col} ---")
    print("Counts:\n", counts)
    print("\nPercentages:\n", percentages.round(2))
    plot_with_percentages(counts, percentages, f"Distribution of {col}", col)

# ===== Step 4: Intersection Plots (Target vs Protected Attributes) =====
target = 'ANXEV_A'
protected_attrs = ['SEX_A', 'RACEALLP_A', 'HISP_A']

for attr in protected_attrs:
    print(f"\n--- Intersection: {target} vs {attr} ---")
    ct = pd.crosstab(df[attr], df[target])
    print(ct)
    print("\nPercentages by row:\n", (ct.div(ct.sum(axis=1), axis=0) * 100).round(2))

    # Plot intersection with percentages on top
    ax = ct.plot(kind='bar', figsize=(7, 5), edgecolor='black')
    plt.title(f"{target} by {attr}")
    plt.xlabel(attr)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.legend(title=target)

    # Add percentage labels on grouped bars
    for p in ax.patches:
        width = p.get_width()
        height = p.get_height()
        x = p.get_x() + width / 2
        if height > 0:
            total_height = ct.sum(axis=1).iloc[int(x // (width * len(ct.columns)))]
            pct = (height / total_height) * 100
            ax.annotate(f"{pct:.1f}%", (x, height), ha='center', va='bottom', fontsize=8, rotation=0)

    plt.show()
```
