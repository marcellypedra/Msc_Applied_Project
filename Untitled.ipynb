```python
#Selecting chosen features for analysis

df = pd.read_csv("adult23.csv", usecols=[
    'ANXEV_A', 'AGEP_A', 'SEX_A', 'RACEALLP_A', 'HISP_A',
    'EDUCP_A', 'MLTFAMFLG_A', 
    'PCNTADLT_A', 'PCNTKIDS_A', 'OVER65FLG_A', 
    'MARSTAT_A', 'CITZNSTP_A'
])
```

```python
print(df.shape)

df.head()
```

```python
#Managing null-vales for the analysis

import numpy as np

invalid_map = {
    'ANXEV_A': [7, 8, 9],
    'AGEP_A': [7, 8, 9],
    'SEX_A': [7, 8, 9],
    'RACEALLP_A': [7, 8, 9],
    'HISP_A': [7, 8, 9],
    'EDUCP_A': [97, 98, 99],  
    'MLTFAMFLG_A': [7, 8, 9],
    'PCNTADLT_A': [7, 8, 9],
    'PCNTKIDS_A': [7, 8, 9],
    'OVER65FLG_A': [7, 8, 9],
    'MARSTAT_A': [9, 97, 98, 99],  
    'CITZNSTP_A': [7, 8, 9]
}

for col, codes in invalid_map.items():
    df[col] = df[col].replace(codes, np.nan)
```

```python
print(df.shape)

df.head()
```

```python
#Checking for Unique values in columns

for col in df.columns:
    print(f"\nUnique values in '{col}':")
    print(df[col].sort_values().unique())
```

```python
#Turning all numbers in integer

exclude_cols = ['MLTFAMFLG_A', 'OVER65FLG_A', 'PCNTADLT_A', 'PCNTKIDS_A', 'HISP_A', 'AGEP_A']

for col in df.columns:
    if col not in exclude_cols:  
        df[col] = df[col].astype('Int64')
```

```python
#Checking for Unique values in columns

for col in df.columns:
    print(f"\nUnique values in '{col}':")
    print(df[col].sort_values().unique())
```

```python
# Checking for missing values

missing_report = df.isnull().sum()

missing_report = missing_report[missing_report > 0]

print("Columns with missing values:")
print(missing_report)
```

```python
# Drop rows with missing data

df.dropna(inplace=True)
```

```python
# evaluating class balance for Anxiety feature

counts = df['ANXEV_A'].value_counts(dropna=False)

percentages = df['ANXEV_A'].value_counts(normalize=True, dropna=False) * 100

print("Class counts:\n", counts)
print("\nClass percentages:\n", percentages)
```

```python
import matplotlib.pyplot as plt

counts.plot(kind='bar', color=['skyblue', 'salmon'])
plt.title("Class Distribution for ANXEV_A (Anxiety Diagnosis)")
plt.xlabel("Class")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Columns to check balance for
cols_to_check = ['ANXEV_A', 'SEX_A', 'RACEALLP_A', 'HISP_A']

for col in cols_to_check:
    print(f"\n--- {col} ---")
    
    # Counts and percentages
    counts = df[col].value_counts()
    percentages = df[col].value_counts(normalize=True) * 100
    
    print("Counts:\n", counts)
    print("\nPercentages:\n", percentages.round(2))
    
    # Plot
    plt.figure(figsize=(6, 4))
    counts.plot(kind='bar', color='skyblue', edgecolor='black')
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()
```

```python
import matplotlib.pyplot as plt

target = 'ANXEV_A'
protected_attrs = ['SEX_A', 'RACEALLP_A', 'HISP_A']

# Individual distributions
for col in [target] + protected_attrs:
    print(f"\n--- {col} ---")
    
    counts = df[col].value_counts()
    percentages = df[col].value_counts(normalize=True) * 100
    
    print("Counts:\n", counts)
    print("\nPercentages:\n", percentages.round(2))
    
    plt.figure(figsize=(6, 4))
    counts.plot(kind='bar', color='skyblue', edgecolor='black')
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

# Intersection plots: ANXEV_A vs each protected attribute
for attr in protected_attrs:
    print(f"\n--- Intersection: {target} vs {attr} ---")
    
    # Create a crosstab
    ct = pd.crosstab(df[attr], df[target])
    print(ct)
    print("\nPercentages by row:\n", (ct.div(ct.sum(axis=1), axis=0) * 100).round(2))
    
    # Plot
    ct.plot(kind='bar', figsize=(7, 5))
    plt.title(f"{target} by {attr}")
    plt.xlabel(attr)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.legend(title=target)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()
```
